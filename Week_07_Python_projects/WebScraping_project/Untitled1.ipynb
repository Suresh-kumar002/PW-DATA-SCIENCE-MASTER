```python
import requests
from bs4 import BeautifulSoup as bs
from urllib.request import urlopen,Request
import logging
```

```python
flipkart_url = "https://www.flipkart.com/search?q=" + "redmi"
flipkart_url
```

```python
headers = {
  "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0 Safari/537.36"
}
```

```python
req = Request(flipkart_url, headers=headers)
```

```python
urlclient = urlopen(req)
urlclient
```

```python
flipkart_page = urlclient.read()
```

```python
flipkart_html = bs(flipkart_page,'html.parser')
```

```python
bigbox = flipkart_html.findAll("div" , {"class":"_1AtVbE col-12-12"})
len(bigbox)
```

```python
bigbox = flipkart_html.findAll("div" , {"class":"_4rR01T"})
len(bigbox)
```

```python
del bigbox[0:3]
```

```python
productlink = "https://www.flipkart.com" + bigbox[3].div.div.div.a['href']
productlink
```

```python
headers = {"User-Agent":"Mozilla/5.0"}
product_req = requests.get(productlink, headers=headers)
```

```python
product_html = bs(product_req.text,'html.parser')
```

```python
review_boxes = product_html.find_all("div", {"class":"_27M-vq"})
len(review_boxes)
```

```python
for review in review_boxes:

    try:
        rating = review.find("div", {"class":"_3LWZlK"}).text
    except:
        rating = "No Rating"

    try:
        title = review.find("p", {"class":"_2-N8zT"}).text
    except:
        title = "No Title"

    try:
        comment = review.find("div", {"class":"t-ZTKy"}).text
    except:
        comment = "No Comment"

    print("Rating:", rating)
    print("Title:", title)
    print("Comment:", comment)
    print("-------------")
```

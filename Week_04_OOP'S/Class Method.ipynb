```python
class pwskills1:
    def __init__(self,name,email):
        self.name=name
        self.email=email
    @classmethod
    def details(cls,name,email):
        return cls(name,email)

    def student_details(self):
        print(self.name,self.email)
```

```python
pw1=pwskills1.details("suru","suru@gmail.com")
```

```python
pw1.name
```

```python
pw1.email
```

```python
pw1.student_details()
```

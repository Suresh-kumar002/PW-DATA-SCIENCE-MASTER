```python
def test():
    print("This is the start of my fun")
    print("This is my fun to test")
    print(4+8)
    print("This is the end of my fun")
```

```python
test()
```

```python
def dec(fun):
    def inner_dec():
        print("This is the start of my fun")
        fun()
        print("This is the end of my fun")
    return inner_dec
```

```python
@dec
def test1():
    print(6+7)
```

```python
test1()
```

```python
import time

def timer_test(fun):
    def timer_test_inner():
        start=time.time()
        fun()
        end=time.time()
        print(end-start)
    return timer_test_inner
```

```python
@timer_test
def test2():
    print(54+78)
```

```python
test2()
```

```python
@timer_test
def test3():
    for i in range(48534687):
        pass
```

```python
test3()
```

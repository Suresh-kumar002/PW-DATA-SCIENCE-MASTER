```python
class pwskill1:

    def __init__(self,phone_no,email_id,student_id):
        self.phone_no=phone_no
        self.email_id=email_id
        self.student_id=student_id

    def return_student_details(self):
        return self.student_id,self.phone_no,self.email_id
```

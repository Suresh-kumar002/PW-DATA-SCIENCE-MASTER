```python
class pwskills2:
    def student_detalis(self,name,email,number):
        print(name,email,number)
        
    @staticmethod
    def mentor_mail_id(mentor_mail_id):
        print(mentor_mail_id)
    
    @staticmethod
    def mentor_class(list_mentor):
        pwskills2.mentor_mail_id(["suru@gmail,com","jugnu@gmail.com"])
        print(list_mentor)

    @classmethod
    def class_name(cls):
        cls.mentor_class(["suru","motii"])

    
    def mentor(self,mentor_list):
        print(mentor_list)
        self.mentor_class(["pagl","motii"])
```

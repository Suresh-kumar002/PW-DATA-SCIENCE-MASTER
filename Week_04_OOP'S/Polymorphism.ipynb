```python
class data_science:
    def syllabus(self):
        print("This is my syllabus for data science master")
```

```python
class web_dev:
    def syllabus(self):
        print("This is my web development syllabus")
```

```python
def class_parecer(class_obj):
    for i in class_obj:
        i.syllabus()
```

```python
data_science=data_science();
```

```python
web_dev=web_dev();
```

```python
class_obj=[data_science,web_dev]
```

```python
class_parecer(class_obj)
```

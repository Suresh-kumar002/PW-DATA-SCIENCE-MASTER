Single Inheritance

```python
class test:

    def test_math(self):
        return "this is my first class"
```

```python
class child_test(test):
    pass
```

```python
child_test_obj=child_test()
```

```python
child_test_obj.test_math()
```

Multilevel Inheritance

```python
class class1:
    def test_class1(self):
        return "This is a from class1"
```

```python
class class2(class1):
    def test_class2(self):
        return "This is a from class2"
```

```python
class class3(class2):
    def test_class3(self):
        return "This is a from class3"
```

```python
class3_obj=class3()
```

```python
class3_obj.test_class1()
```

```python
class3_obj.test_class2()
```

```python
class3_obj.test_class3()
```

MULTIPLE INHERITANCE

```python
class class1:
    def test_class1(self):
        return "This is a from class1"
```

```python
class class2:
    def test_class2(self):
        return "This is a from class2"
```

```python
class class3(class1,class2):
    def test_class3(self):
        return "This is a from class3"
```

```python
class3_obj=class3()
```

```python
class3_obj.test_class1()
```

```python
class3_obj.test_class2()
```

Hierarchical Inheritance

```python
class class1:
    def test_class1(self):
        return "This is s class1"
```

```python
class class2(class1):
    def test_class2(self):
        return "This is a class2"
```

```python
class class3(class1):
    def test_class3(self):
        return "This is a class3"
```

```python
class3_obj=class3()
```

```python
class3_obj.test_class1()
```

```python
class2_obj=class2()
```

```python
class2_obj.test_class1()
```

Hybrid Inheritance

```python
class class1:
    def test_class1(self):
        return "This is a from class1"
```

```python
class class2(class1):
    def test_class2(self):
        return "This is a from class2"
```

```python
class class3(class1):
    def test_class3(self):
        return "This is a from class3"
```

```python
class class4(class2,class3):
    def test_class4(self):
        return "This is a from class4"
```

```python
class4_obj=class4()
```

```python
class4_obj.test_class1()
```

```python
class4_obj.test_class2()
```

```python
class4_obj.test_class3()
```

```python
class4_obj.test_class4()
```

```python
class bank_acc:
    def __init__(self,balance):
        self.__balance=balance

    def deposit(self,amount):
        self.__balance=self.__balance+amount

    def withdaw(self,amount):
        if self.__balance>=amount:
            self.__balance=self.__balance-amount
            return True
        else:
            return False

    def get_balance(self):
        return self.__balance
        
```

```python
suru=bank_acc(467)
```

```python
suru.get_balance()
```

```python
suru.deposit(1275)
```

```python
suru.get_balance()
```

```python
suru.withdaw(1000)
```

```python
suru.get_balance()
```

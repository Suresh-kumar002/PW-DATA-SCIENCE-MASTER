```python
#use always a specific exception
try:
    10/0
except Exception as e:
    print(e)
```

```python
try:
    10/0
except ZeroDivisionError as e:
    print(e)
```

```python
#print always a proper msg
try:
    10/0
except ZeroDivisionError as e:
    print("I am trying to handel a zerodivision error",e)
```

```python
#always try to log your error
import logging
logging.basicConfig(filename="error.log",level=logging.ERROR)
try:
    10/0
except ZeroDivisionError as e:
    logging.error("I am trying to handel a zerodivision error {}".format(e))
```

```python
#always avoid to write multiple exception handling
try:
    10/0
except FileNotFoundError as e:
    logging.error("I am handeling file not found {}".format(e))
except AttributeError as e:
    logging.error("I am handeling Attribute error {}".format(e))
except ZeroDivisionError as e:
    logging.error("I am trying to handel a zerodivision error {}".format(e))
```

```python
#Document all the error
#cleanup all the resources
try:
    with open("test.txt","w") as f:
        f.write("this is my data to file")
except FileNotFoundError as e:
    logging.error("I am handeling file not found {}".format(e))
finally:
    f.close()
```

```python
import multiprocessing
def square(n):
    return n**2

if __name__ == '__main__':
    with multiprocessing.Pool(processes=4) as pool:
        out=pool.map(square,[1,2,3,4,5,6,67,78,7])
        print(out)
```

```python
def producer(q):
    for i in range(10):
        q.put(i)

def consumer(q):
    while True:
        item=q.get()
        if item is None:
            break
    print(item)

if __name__ =='__main__':
        queue=multiprocessing.Queue()
        m1=multiprocessing.Process(target=producer,args=(queue,))
        m2=multiprocessing.Process(target=consumer,args=(queue,))
        m1.start()
        m2.start()
        queue.put("suru")
        m1.join()
        m2.join()
```

```python
import multiprocessing
def producer(q):
    for i in ["suru","tanwar","moti"]:
        q.put(i)

def consumer(q):
    while True:
        item=q.get()
        if item is None:
            break
    print(item)

if __name__ =='__main__':
        queue=multiprocessing.Queue()
        m1=multiprocessing.Process(target=producer,args=(queue,))
        m2=multiprocessing.Process(target=consumer,args=(queue,))
        m1.start()
        m2.start()
        queue.put("suru")
        m1.join()
        m2.join()
```

```python
import multiprocessing
def square(index , value) :
    value[index] = value[index]**2
    
if __name__ == '__main__':
    arr = multiprocessing.Array('i' , [2,3,4,5,6,7,8])
    process = []
    for i in range(7):
        m = multiprocessing.Process(target=square , args = (i,arr))
        process.append(m)
        m.start()
    for m in process:
        m.join()
    print(list(arr))
```

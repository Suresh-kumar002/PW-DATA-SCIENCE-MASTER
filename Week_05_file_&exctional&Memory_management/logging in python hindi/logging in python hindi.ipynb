```python
l=[1,2,3,4,545,654,5,67,[3,4,3,5,3,4],"suresh","kumar"]
```

```python
lst_int=[]
lst_str=[]
for i in l:
    logging.info("we are iterating through our list and our local var is i {}".format(l))
    if type(i)==list:
        logging.info("i am inside if statement and i am trying to check list type"+str(i))
        for j in i:
            logging.info("i am in another for loop for list inside list element"+str(j))
            if type(i)==int:
                logging.info("i am inside if statement")
                lst_int.append(j)
    elif type(i)==int:
        lst_int.append(i)
    else:
        if type(i)==str:
            lst_str.append(i)
logging.info("my final result for int is {l1} and str is {l2}".format(l1=lst_int,l2=lst_str))
```

```python
import time
```

```python
def test1(id):
    for i in range(10):
        print("test1 %d printing %d %s" %(id,i,time.ctime()))
        time.sleep(1)
```

```python
test1(1)
```

```python
thread1=[threading.Thread(target=test1,args=(i,)) for i in range(3)]
```

```python
thread1
```

```python
for t in thread1:
    t.start()
```
